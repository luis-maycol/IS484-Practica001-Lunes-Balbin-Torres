<h1 style="text-align: center; color: #1a5276;">Sistema Experto de Soporte Técnico (Nivel 1)</h1>
<p style="text-align: justify; font-size: 13px; color: #333333;">
Agente lógico conversacional diseñado para diagnosticar y resolver problemas comunes de hardware y software en equipos de cómputo, derivando a soporte especializado cuando el nivel de severidad lo amerite.
</p>

<h2><span style="color: #2980b9;">1. Diseño del Agente: Ficha PEAS</span></h2>

<h2><span style="color: #2980b9;">2. Justificación de las reglas de decisión</span></h2>
<ol style="font-size: 13px; color: #333333; line-height: 1.6; margin-left: 20px;">
  <li><b>Problemas de encendido:</b> Si el equipo no enciende, se deriva inmediatamente a <i>Nivel 2</i> (hardware), ya que no es posible una intervención por software remoto.</li>
  <li><b>Errores críticos (Pantalla azul/BSOD):</b> Se clasifica como fallo grave. La acción depende de si el usuario tiene respaldo de datos; si no lo tiene, la prioridad sube para evitar pérdida de información.</li>
  <li><b>Lentitud o problemas de red:</b> Son las incidencias más comunes. Se resuelve con instrucciones automatizadas de <i>Nivel 1</i> (limpieza de temporales, reinicio de enrutador).</li>
  <li><b>Síntomas desconocidos:</b> El agente pide reformular o dar más detalles si no detecta palabras clave en la descripción del problema.</li>
</ol>

```python
def agente_soporte_ti(sintoma_reportado, equipo_enciende, tiene_respaldo):
    """
    Agente reactivo simple que clasifica y da solucion a incidentes de TI.
    Retorna una tupla: (accion, recomendacion)
    """
    sintoma = sintoma_reportado.lower()
    
    # Regla 1: Hardware físico
    if not equipo_enciende:
        return "Derivar a Nivel 2", "Fallo fisico (fuente de poder o placa). Se requiere tecnico presencial."
    
    # Regla 2: Fallo crítico de sistema
    if "pantalla azul" in sintoma or "bsod" in sintoma:
        if tiene_respaldo:
            return "Derivar a Nivel 2", "Sistema corrupto. Al existir respaldo de datos, el tecnico procedera a formatear."
        else:
            return "ALERTA Nivel 2", "Fallo critico sin respaldo. Apagar equipo para intentar recuperar disco duro en laboratorio."
            
    # Regla 3: Mantenimiento básico
    if "lento" in sintoma or "congela" in sintoma:
        return "Solucion Nivel 1", "Por favor elimine archivos temporales y ejecute el antivirus. Cierre procesos en segundo plano."
        
    # Regla 4: Conectividad
    if "internet" in sintoma or "red" in sintoma or "wifi" in sintoma:
        return "Solucion Nivel 1", "Reinicie su router y verifique que el cable de red este bien conectado al equipo."
        
    # Regla por defecto (Fallback)
    return "Pedir mas informacion", "No comprendo el fallo exacto. ¿Podria indicar si aparece algun codigo de error en pantalla?"

# Matriz de Pruebas
casos_prueba_soporte = [
    ("Mi laptop no hace nada al presionar el boton", False, True),           # Caso 1: No enciende
    ("Sale una pantalla azul con letras blancas", True, False),             # Caso 2: BSOD sin respaldo
    ("Ayer hubo un pantallazo azul pero ya prendio", True, True),           # Caso 3: BSOD con respaldo
    ("El sistema operativo esta muy lento", True, True),                    # Caso 4: Lentitud
    ("No me puedo conectar a la red WiFi de la oficina", True, False),      # Caso 5: Red
    ("No puedo imprimir en la impresora 3D", True, True)                    # Caso 6: No contemplado (Fallback)
]

print(f"{'N°':<3} | {'Acción del Agente':<20} | {'Recomendación / Diagnóstico'}")
print("-" * 115)
for i, (sintoma, enciende, respaldo) in enumerate(casos_prueba_soporte, 1):
    accion, recomendacion = agente_soporte_ti(sintoma, enciende, respaldo)
    print(f"{i:<3} | {accion:<20} | {recomendacion}")
```
