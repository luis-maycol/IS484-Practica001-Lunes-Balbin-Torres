<h1 style="text-align: center; color: #1a5276;">Agente de Control de Calidad en Ensamblaje</h1>
<p style="text-align: justify; font-size: 13px; color: #333333;">
Sistema inteligente reactivo para la inspección automatizada de piezas en una línea de producción, evaluando tolerancias físicas y defectos visuales para decidir la aprobación del lote.
</p>

<h2><span style="color: #2980b9;">1. Diseño del Agente: Ficha PEAS</span></h2>

<h2><span style="color: #2980b9;">2. Justificación de las reglas de decisión</span></h2>
<p style="text-align: justify; font-size: 13px; color: #333333;">Parámetros objetivo: Peso ideal de 100g y Longitud ideal de 10.0cm.</p>
<ol style="font-size: 13px; color: #333333; line-height: 1.6; margin-left: 20px;">
  <li><b>Defecto visual detectado:</b> Genera un rechazo inmediato de la pieza independientemente de sus medidas físicas, por incumplimiento de estándares estéticos o daños estructurales.</li>
  <li><b>Medidas fuera de tolerancia crítica:</b> Si el peso se desvía más de ±5g (fuera de 95g-105g) o la longitud varía más de ±0.5cm (fuera de 9.5cm-10.5cm), la pieza es rechazada por ser disfuncional.</li>
  <li><b>Medidas en límite de tolerancia:</b> Si las medidas son aceptables pero rozan los extremos permitidos (ej. peso entre 95-96g o 104-105g), la pieza se aprueba, pero se emite una <i>alerta de calibración</i> para que los técnicos ajusten la maquinaria antes de generar defectos reales.</li>
  <li><b>Medidas óptimas:</b> Aprobación directa sin observaciones.</li>
</ol>

<h2><span style="color: #2980b9;">3. Código del Agente Inteligente</span></h2>

```python
def agente_inspeccion(peso, longitud, defecto_visual):
    """
    Evalua piezas en una linea de produccion basado en tolerancias fisicas.
    Retorna una tupla: (accion, motivo)
    """
    # 1. Regla de defecto visual (Prioridad Alta)
    if defecto_visual:
        return "RECHAZAR", "Anomalia visual detectada en la superficie"
        
    # 2. Regla de tolerancia critica
    if peso < 95.0 or peso > 105.0:
        return "RECHAZAR", f"Peso fuera de especificacion ({peso}g)"
    if longitud < 9.5 or longitud > 10.5:
        return "RECHAZAR", f"Longitud fuera de especificacion ({longitud}cm)"
        
    # 3. Regla de alerta preventiva (Cercano a los limites)
    peso_limite = (95.0 <= peso <= 96.0) or (104.0 <= peso <= 105.0)
    longitud_limite = (9.5 <= longitud <= 9.6) or (10.4 <= longitud <= 10.5)
    
    if peso_limite or longitud_limite:
        return "APROBAR - ALERTA", "Pieza valida pero medidas en el limite. Revisar calibracion de maquinas"
        
    # 4. Regla optima
    return "APROBAR", "Medidas dentro del rango optimo"

# Matriz de Pruebas
print("<h2><span style='color: #2980b9;'>4. Matriz de Pruebas y Resultados</span></h2>")

casos_prueba_calidad = [
    (100.0, 10.0, False),  # Caso 1: Pieza perfecta
    (95.5,  10.4, False),  # Caso 2: Alerta en ambos limites (pero valido)
    (106.0, 10.0, False),  # Caso 3: Rechazo por sobrepeso
    (100.0, 9.2,  False),  # Caso 4: Rechazo por longitud muy corta
    (100.0, 10.0, True),   # Caso 5: Rechazo por defecto visual (medidas perfectas)
    (96.5,  9.8,  False)   # Caso 6: Pieza aprobada, ligera variacion segura
]

print(f"{'N°':<3} | {'Peso(g)':<7} | {'Long(cm)':<8} | {'Visual':<6} | {'Acción':<18} | {'Motivo'}")
print("-" * 105)
for i, (p, l, v) in enumerate(casos_prueba_calidad, 1):
    accion, motivo = agente_inspeccion(p, l, v)
    print(f"{i:<3} | {p:<7} | {l:<8} | {str(v):<6} | {accion:<18} | {motivo}")
```
